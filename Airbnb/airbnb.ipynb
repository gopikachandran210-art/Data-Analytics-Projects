```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('datasets.csv', encoding_errors='ignore')
```

```python
Data Cleaning
```

```python
data.isnull().sum()
# dropping missing values
data.dropna(inplace = True)
data.isnull().sum()
```

```python
data.shape
```

```python
# dealing with duplicates row
data.duplicated().sum()
# deleting all duplicated rows
# data[data.duplicated()]

data.drop_duplicates(inplace=True)
data[data.duplicated()]
```

```python
data.dtypes
```

```python
# type cating 
# changing the type of id host_is as keeping it int or float we dont need to do any numerical operations
data.dtypes
data['id']=data['id'].astype(object)
data.dtypes
data['host_id']=data['host_id'].astype(object)
data.dtypes
```

```python
Data Analysis
*****EDA*****
    **Univariant Analysis**(Analysis of each column spread of the values)
```

```python
# Identify outliers
df = data[data['price'] < 1500]
sns.boxplot(data=df, x='price')
```

```python
# Price disrtibution
# data['price']
plt.figure(figsize=(8,5))
sns.histplot(data=df, x = 'price', bins=100)
plt.title('Price distribution')
plt.ylabel('Frequency')
plt.show()
```

```python
df.columns
```

```python
# Availability Distribution
plt.figure(figsize=(6,3))
sns.histplot(data=df, x = 'availability_365')
plt.title('availability_365 Distribution')
plt.ylabel('Frequency')
plt.show()
```

```python
data.dtypes
```

```python
df.groupby(by = 'neighbourhood_group')['price'].mean()
```

```python
# average price per bed
df.groupby(by = 'neighbourhood_group')['price per bed'].mean()
```

```python
**Feature Engineering**
```

```python
# Price per bed
df['price per bed'] = df['price']/df['beds']
df.head()
```

```python
**Bi variable Analysis**
One variable dependency in other variable
```

```python
# price cdependency and neighbourhood grop
sns.barplot(data = df, x = 'neighbourhood_group', y = 'price', hue = 'room_type')
```

```python
# number of reviews 
plt.figure(figsize = (8,5))
plt.title('Locality and Review dependency')
sns.scatterplot(data=df, x = 'number_of_reviews', y ='price', hue = 'neighbourhood_group')
plt.show()
```

```python
df.dtypes
```

```python
sns.pairplot(df, vars = ['price', 'minimum_nights', 'number_of_reviews', 'availability_365'], hue = 'room_type')
```

```python
# Geographical distribution airbnb 
plt.figure(figsize=(10,7))
sns.scatterplot(data= df, x ='longitude', y ='latitude', hue='room_type')
plt.title('Geographical Distribution of Airbnb Listing')
plt.show()
```

```python
# heat map chart - correlation of one variable with other for numerical columsn
corr =df[['latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month', 'availability_365', 'beds']].corr()
corr
plt.figure(figsize=(8,6))
sns.heatmap(data= corr, annot = True)
```
